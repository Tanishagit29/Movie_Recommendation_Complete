## 1. Import libraries and define project folders


````python
from pathlib import Path
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
````

## 2. Download MovieLens 1M

Run `train.py` for the complete automated pipeline. It downloads `ml-1m.zip` from GroupLens and parses `ratings.dat` and `movies.dat`.


````python
print("Run from VS Code: python train.py")
````

## 3. Data loading and cleaning

Ratings contain userId, movieId, rating and timestamp. Movies contain movieId, title and genres. Ratings are sorted chronologically.


## 4. EDA

Inspect rating distribution, missing values, duplicates, number of users, movies and ratings.


## 5. Relevance definition

For ranking, rating >= 4 is treated as relevant (1); rating < 4 is not relevant (0).


## 6. Temporal train/test split

The first 80% of interactions are training data and the final 20% are held-out test interactions.


## 7. Feature engineering

User statistics, movie statistics, genre indicators, user genre preferences, popularity, rating difference and genre-match score are generated from training data.


## 8. Negative sampling

Unseen movies are sampled as negative examples so the binary classifiers learn relevant vs non-relevant interactions.


## 9. Baseline

A Bayesian weighted movie popularity score provides the non-personalized baseline.


## 10. XGBoost

XGBoost is trained as the first boosting classifier.


## 11. LightGBM

LightGBM is trained as the second boosting classifier.


## 12. Leakage-safe stacking

Five-fold out-of-fold predictions from XGBoost and LightGBM train a Logistic Regression meta learner.


## 13. Classification evaluation

Accuracy, Precision, Recall, F1 and ROC-AUC are reported as secondary diagnostics.


## 14. Ranking evaluation

For each user, held-out relevant movies are ranked with sampled unseen negatives. Precision@10, Recall@10 and NDCG@10 are averaged across users.


## 15. Model comparison

`outputs/model_comparison.csv` and `outputs/ranking_comparison.csv` contain the results.


## 16. Explainability

XGBoost and LightGBM feature importance are saved. SHAP summary is generated for XGBoost when SHAP is available.


## 17. Deployment

All model and feature artifacts are saved under `models/`. Run `streamlit run app.py` to launch the recommendation application.


## 18. Final conclusion

The final report should emphasize the recommendation/ranking metrics and explain the difference between the popularity baseline, individual boosting models and the stacking ensemble.
